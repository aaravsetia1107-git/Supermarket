```python
# Cell 1
Electronics={'Phone':5, 'Laptop':7, 'Headphones':4, 'Earbuds':8}
Fruits_and_vegetables={'Apples':7, 'Banana':8, 'Watermelon':2, 'Mango':12}
Groceries={'Bread':5, 'Flour':9, 'Juice':4, 'Paneer':8}
Self_care={'Shampoo':6, 'Conditioner':6, 'Face Wash':9}
Supermarket=[Electronics, Fruits_and_vegetables, Groceries, Self_care]
names=['Electronics', 'Fruits and Vegetables', 'Groceries', 'Self Care']

cart={}
```

```python
# Cell 2
while True:
    print()
    print('''What Do you want to do?
1) Shop
2) Return Something
3) Checkout
4) Quit''')
    n=input('Enter your choice: ')
    if not n.isdigit():
        print('Please enter a number between 1 and 4')
        continue
    n=int(n)

    if n==1:
        while True:
            print()
            for i in range(4):
                print(i+1, names[i])
            print('5 Go back')
            c=input('Choose a category: ')
            if not c.isdigit():
                print('Wrong input')
                continue
            c=int(c)
            if c==5:
                break
            if c<1 or c>4:
                print('No such category')
                continue

            cat=Supermarket[c-1]
            items=list(cat)
            print()
            for i in range(len(items)):
                print(i+1, items[i], '- Rs', cat[items[i]])
            p=input('Enter item number (0 to go back): ')
            if not p.isdigit():
                print('Wrong input')
                continue
            p=int(p)
            if p==0:
                continue
            if p>len(items):
                print('No such item')
                continue

            q=input('How many do you want? ')
            if not q.isdigit() or int(q)==0:
                print('Quantity should be a number greater than 0')
                continue
            q=int(q)
            item=items[p-1]
            if item in cart:
                cart[item][1]+=q
            else:
                cart[item]=[cat[item], q]
            print(item, 'added to cart')

    elif n==2:
        if len(cart)==0:
            print('Your cart is empty, nothing to return')
        else:
            print()
            things=list(cart)
            for i in range(len(things)):
                print(i+1, things[i], '- Rs', cart[things[i]][0], 'x', cart[things[i]][1])
            r=input('Enter item number to return (0 to go back): ')
            if not r.isdigit():
                print('Wrong input')
            elif int(r)==0:
                print('Going back')
            elif int(r)>len(things):
                print('No such item in cart')
            else:
                item=things[int(r)-1]
                q=input('How many do you want to return? ')
                if not q.isdigit() or int(q)==0:
                    print('Quantity should be a number greater than 0')
                elif int(q)>=cart[item][1]:
                    del cart[item]
                    print(item, 'removed from cart')
                else:
                    cart[item][1]-=int(q)
                    print(q, item, 'returned')

    elif n==3:
        if len(cart)==0:
            print('Your cart is empty, shop something first')
        else:
            print()
            print('      BILL      ')
            total=0
            for item in cart:
                price=cart[item][0]
                qty=cart[item][1]
                print(item, ': Rs', price, 'x', qty, '= Rs', price*qty)
                total+=price*qty
            print('                ')
            print('Total = Rs', total)
            pay=input('Pay now? (y/n): ')
            if pay=='y' or pay=='Y':
                print('Payment done. Thank you for shopping!')
                cart={}
            else:
                print('Payment cancelled, your cart is still saved')

    elif n==4:
        print('Thank you, visit again!')
        break

    else:
        print('Please enter a number between 1 and 4')
```
